# Data Augmentation
### We can clearly see that our model was underfitting since the loss from validation and testing is same.

```python
import numpy as np
```

```python
data = np.load('non-correlated_files.npy')
data.shape
```

## Augmentation Technique #1
### Frame Splits

```python
def printAP(a,d,n): 
    curr_term=a 
    series=[-1]
    for i in range(1,n+1): 
        series.append(curr_term)
        curr_term =curr_term + d 
    return series

a = 14 # starting number 
d = 15 # Common difference 
n = 9 # N th term to be find 

seggregator=printAP(a, d, n)
print(printAP(a, d, n) )
```

```python
frames=[]
for i in range(0,73):
    for j in range(0,9):
        frames.append(np.concatenate((data[i,seggregator[j]+1:seggregator[j+1],:,:,:],data[i,135:141,:,:,:]),axis=0))

frames=np.array(frames) 
```

```python
np.save('data_with_frame_splits', frames)
frames.shape
```

## Augmentation Technique #2
### Frame Shifts

```python
data = np.load('non-correlated_files.npy')
data.shape
```

```python
frames=[]
for i in range(0,73):
    for j in range(0,120):
        frames.append(np.concatenate((data[i,j:j+15,:,:,:],data[i,135:141,:,:,:]),axis=0))

frames=np.array(frames)
frames.shape
```

```python
np.save('data_with_frame_shifts', frames)
frames.shape
```
